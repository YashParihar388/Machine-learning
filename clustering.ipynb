```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
```

```python
x,y_true=make_blobs(n_samples=500,centers=3,cluster_std=4,random_state=42)
```

```python
df = pd.DataFrame(x,columns=['feature1','feature2'])
```

```python
le = StandardScaler()
x_scaled = le.fit_transform(df)
```

```python
k_range = range(1,11)
inertia = []
```

```python
for k in k_range:
    kmeans =KMeans(n_clusters=k,random_state=42)
    model= kmeans.fit(x_scaled)
    inertia.append(model.inertia_)
```

```python
plt.plot(k_range,inertia,marker='o')
```

```python
k_final = KMeans(n_clusters=3,random_state=42)
```

```python
clusters = k_final.fit_predict(x_scaled)
```

```python

df['cluster'] = clusters
```

```python
sns.scatterplot(
    x=df['feature1'],
    y=df['feature2'],
    hue=df['cluster'],
    palette='viridis'
)
```

```python
from sklearn.datasets import make_moons
```

```python
x,y_true = make_moons(n_samples=500,random_state=42,noise=0.5)
```

```python
from sklearn.cluster import KMeans, DBSCAN
```

```python

df = pd.DataFrame(x, columns=['Feature_1', 'Feature_2'])
```

```python

scaler = StandardScaler()
X_scaled = scaler.fit_transform(df)
```

```python
kmeans = KMeans(n_clusters=2, random_state=42)
kmeans_labels = kmeans.fit_predict(X_scaled)
```

```python

df['kmeans_cluster'] = kmeans_labels
```

```python

sns.scatterplot(x=df['Feature_1'], y=df['Feature_2'], hue=df['kmeans_cluster'], palette='tab10')
```

```python
dbscan = DBSCAN(eps=0.3,min_samples=5)
dbscan_labels = dbscan.fit_predict(X_scaled)
```

```python

df['dbscan_cluster'] = dbscan_labels
```

```python
sns.scatterplot(
    x=df['Feature_1'],
    y=df['Feature_2'],
    hue=df['dbscan_cluster'],
    palette='tab10'
)
```
